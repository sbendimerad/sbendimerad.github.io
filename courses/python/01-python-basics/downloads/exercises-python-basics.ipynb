## Exercise 9 — Seven-Segment Display 


**Estimated Time:** 45 minutes  
**Level of Difficulty:** Hard  


**Scenario**  
You’ve surely seen a seven-segment display: a device that shows one decimal digit by lighting a subset of seven segments.  
In this exercise, you’ll **simulate** such a display using **single LEDs** (characters) instead of physical segments.

Each digit is built from **13 LEDs** arranged on **5 text rows** (some lit, some dark). Here’s how we imagine the output:


```
  # ### ### # # ### ### ### ### ### ### 
  #   #   # # # #   #     # # # # # # # 
  # ### ### ### ### ###   # ### ### # # 
  # #     #   #   # # #   # # #   # # # 
  # ### ###   # ### ###   # ### ### ###
```

> Note: digit **8** lights all LEDs.

**Task**  
Write a function `print_number(num)` that prints any **non-negative integer** using the LED layout above.

**Tip**  
Use a list (or dict) of 10 patterns (one per digit). For example:

```python
digits = [
    '1111110',  # 0
    '0110000',  # 1
    '1101101',  # 2
    '1111001',  # 3
    '0110011',  # 4
    '1011011',  # 5
    '1011111',  # 6
    '1110000',  # 7
    '1111111',  # 8
    '1111011',  # 9
]
```


**Hints:**
- Think in 5 text rows × 3 columns per digit, then concatenate across digits.
- Map each of the 7 segments (indexes 0–6 in the pattern string) to the LED positions you want to turn on.
- Build each row as a string and print at the end to avoid flicker/misalignment.
