### Loops

- `for` on sequences / `range()`
- `enumerate()` for index + value
- `break` / `continue`

💡 **Tip**: use `break` to stop a loop and `continue` to skip to the next iteration.

```python
for i in range(5):
    print("Iteration", i)
```

```python
for i in range(10):
    if i % 2 == 0:   # if i is even
        continue     # skip directly to the next iteration
    print(i)
```

```python
names = ["Ana", "Bob", "Chloe"]
collected = []
for i, name in enumerate(names):
    if name.startswith("B"):
        continue
    collected.append((i, name.upper()))
collected
```

```python
# while loop: pay attention to the stopping condition
x = 0
res = []
while True:
    res.append(x)
    x += 1
    if x >= 5:
        break
res
```

#### **Activity: Even and Odd Number Counter (15 min)**

👉 **Objective**: practice loops, conditions, and counters.

Write a Python program that allows the user to enter a series of numbers until they enter `0` to stop the execution.  

The program must:  

- Count how many **even** and **odd** numbers were entered  
- Display these results at the end  

💡 **Instructions**  
Implement this program step by step:  

1. Create two counters: `evens` and `odds`.  
2. Use a `while` loop to ask the user for a number.  
3. Check if this number is even or odd.  
4. Increment the corresponding counter.  
5. Stop when the user enters `0`.  
6. Display the final result.  


```python
# ask the user to enter a number (input)
# keep asking for numbers until the user enters 0 (while)
# check whether it is even or odd (if and modulo)
# if it is even -> add it to the even counter
# otherwise -> add it to the odd counter 
```

```python
odd_numbers = 0
even_numbers = 0
number = int(input("Enter a number or type 0 to stop: "))

while number != 0:
    # check if the number is odd
    if number % 2 == 1:
        # increase the odd_numbers counter
        odd_numbers += 1
    else:
        # increase the even_numbers counter
        even_numbers += 1
    # read the next number
    number = int(input("Enter a number or type 0 to stop: "))

# print results
print("Odd numbers count:", odd_numbers)
print("Even numbers count:", even_numbers)
```

#### **Activity: Bubble Sort (20 min)**
👉 Sorting algorithms — Bubble Sort

- Sorting a sequence of 10 numbers doesn’t take long. But what about when we have **100,000 numbers** to sort?  
- That’s when we use **sorting algorithms**.  

Sorting algorithms are the foundation of algorithmics and allow us to reorganize our data.  
Several algorithms exist, but here we focus on **Bubble Sort** because it is:  

- Easy to understand  
- Useful for sorting lists (though very resource intensive)  

### Why the name "Bubble Sort"?
During sorting, the largest element gradually moves to the end of the list, like a **bubble** rising to the surface.

### Principle of the algorithm

- Move forward through the list  
- Compare elements **two by two**  
- Swap them if `value1 > value2`  
- Repeat until the list is sorted  

Example with `[8, 10, 6, 2, 4]`:  

- Step 1: compare 8 and 10 → nothing changes  
- Step 2: compare 10 and 6 → swap → `[8, 6, 10, 2, 4]`  
- Step 3: compare 10 and 2 → swap → `[8, 6, 2, 10, 4]`  
- Step 4: compare 10 and 4 → swap → `[8, 6, 2, 4, 10]`  

💡 **Instructions**  
Implement the **Bubble Sort** algorithm yourself to sort a list of numbers.



```python
# Example: using range in a for loop
myList = [10, 1, 8, 3, 5]
swapped = True # a little fake, need it to enter the while loop

while swapped:
    swapped = False # no swaps
    for i in range(len(myList) - 1):
        if myList[i] > myList[i + 1]:
            swapped = True # swap occured!
            myList[i], myList[i + 1] = myList[i + 1], myList[i]

print(myList)
```

```python
def normalize_price(s):
    """Convert '£51.77' -> 51.77 (float)."""
    s = s.strip().replace("£", "").replace(",", ".")
    return float(s)

normalize_price("£51.77")
```

```python
# Function definition
# Early return & reuse

def age_category(age):
    if age < 0:
        return "invalid"
    if age < 12:
        return "child"
    if age < 18:
        return "teenager"
    return "adult"

[age_category(a) for a in (-1, 5, 16, 30)]
```

## 11) Exceptions (overview)

```python
# 🏦 Example 1 – ATM Withdrawal

def withdraw(balance, amount):
    try:
        return balance - amount
    except TypeError:
        print("Error: amount must be a number!")

print(withdraw(100, 50))     # works → 50
print(withdraw(100, "fifty"))# error handled
```

```python
# 🚖 Example 2 – Taxi Price Calculation

def price_per_km(distance, price):
    try:
        return price / distance
    except ZeroDivisionError:
        print("Error: distance cannot be zero!")
    except TypeError:
        print("Error: distance and price must be numbers")

print(price_per_km(10, 20))   # → 2.0 €/km
print(price_per_km(0, 20))    # distance zero → error handled
print(price_per_km("ten", 20))# wrong input type → error handled
```

```python
# 🛒 Example 3 – Online Order

def get_price(cart, item):
    try:
        return cart[item]
    except KeyError:
        print(f"Error: '{item}' not found in cart!")

cart = {"apple": 1.5, "banana": 2.0}
print(get_price(cart, "apple"))   # works → 1.5
print(get_price(cart, "orange"))  # KeyError caught
```

```python
# 🧩 Example – User enters unknown input

def to_float(user_input):
    try:
        return float(user_input.replace(",", "."))
    except Exception as e:  
        print("Something went wrong:", e)

print(to_float("12.5"))   # works → 12.5
print(to_float("7,5"))    # works → 7.5
print(to_float("oops"))   # catches ValueError → prints message
print(to_float(None))     # catches AttributeError → prints message
```
