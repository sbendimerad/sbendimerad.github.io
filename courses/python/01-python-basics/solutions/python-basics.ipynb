## Exercise 2 — Collatz Conjecture {#ex-exercise-2-collatz-conjecture}

**Estimated Time:** 10 minutes  
**Level of Difficulty:** Medium  

**Task:**

- In 1937, Lothar Collatz formulated a hypothesis:
  - take any positive integer c0;
  - if even → c0 = c0 ÷ 2;
  - if odd → c0 = 3 × c0 + 1;
  - repeat until c0 = 1.

- Write a program to apply these rules and print each value of c0.
- Count and print the number of steps.

**Test Data:**

- Input: 16 → Output: 8, 4, 2, 1; Steps = 4

````python
c0 = int(input("Enter Starting Number: "))

counter = 0

while c0 != 1:
    counter += 1
    if c0 % 2 == 0:
        c0 = c0 // 2
        print(c0)
    else:
        c0 = (3 * c0) + 1
        print(c0)

print("Steps =", counter)
````

## Exercise 4 — Leap Year Checker {#ex-exercise-4-leap-year-checker}

**Estimated Time:** 15 minutes  
**Level of Difficulty:** Easy  

**Task:**

- Determine if a year is a leap year, common year, or not within the Gregorian calendar (before 1582).

**Rules:**

- Leap year if divisible by 4;
- Except divisible by 100;
- Unless divisible by 400.

<img src="../img/Leapyear.jpeg" alt="Leap year rules" width="300">

**Test Data:**

- Input: 2000 → Leap year
- Input: 2015 → Common year
- Input: 1999 → Common year
- Input: 1996 → Leap year
- Input: 1580 → Not within Gregorian calendar

````python
year = int(input("Enter a year: "))

if year >= 1582:
    if year % 4 == 0:
        if year % 100 == 0:
            if year % 400 == 0:
                print("Leap year")
            else:
                print("Common year")
        else:
            print("Leap year")
    else:
        print("Common year")
else:
    print("Not within the Gregorian calendar period")
````

## Exercise 7 — Pyramid Blocks {#ex-exercise-7-pyramid-blocks}

**Estimated Time:** 15 minutes  
**Level of Difficulty:** Medium  

**Task:**

- Write a program which reads the number of blocks available.
- Output the maximum height of the pyramid that can be built.
- Each layer uses one more block than the layer above.
- Stop when not enough blocks remain to build the next layer.

<img src="../img/pyramid_blocks.png" alt="Pyramid of blocks" width="300">

**Example:**

- Input: 6 → Output: Height = 3

````python
blocks = int(input("Enter the number of blocks: "))

height = 0
inlayer = 1

while blocks >= inlayer:
    height += 1
    blocks -= inlayer
    inlayer += 1

print("The height of the pyramid:", height)
````

````python
from random import choice

def display_board(board):
    # Displays the current state of the board
    print("+-------" * 3, "+", sep="")
    for row in range(3):
        print("|       " * 3, "|", sep="")
        for col in range(3):
            print("|   " + str(board[row][col]) + "   ", end="")
        print("|")
        print("|       " * 3, "|", sep="")
        print("+-------" * 3, "+", sep="")

def enter_move(board,free_fields):
    # Allows the player to enter their move, checks for validity, and updates the board
    move = int(input("Enter your move (1-9): "))
    row = (move - 1) // 3
    col = (move - 1) % 3
    while (row, col) not in free_fields:
        print("Invalid move. The position is either occupied or out of range.")
        move = int(input("Enter your move (1-9): "))
        row = (move - 1) // 3
        col = (move - 1) % 3
    board[row][col] = 'O'
    return board

def make_list_of_free_fields(board):
    # Returns a list of all unoccupied positions on the board
    free_fields = []
    for row in range(3):
        for col in range(3):
            if isinstance(board[row][col], int):
                free_fields.append((row, col))
    return free_fields

def victory_for(board, sign):
    # Checks if there's a winning condition on the board for the given sign ('X' or 'O')
    win_conditions = [
        [(0, 0), (0, 1), (0, 2)], [(1, 0), (1, 1), (1, 2)], [(2, 0), (2, 1), (2, 2)],
        [(0, 0), (1, 0), (2, 0)], [(0, 1), (1, 1), (2, 1)], [(0, 2), (1, 2), (2, 2)],
        [(0, 0), (1, 1), (2, 2)], [(0, 2), (1, 1), (2, 0)]
    ]
    for condition in win_conditions:
        if all(board[row][col] == sign for row, col in condition):
            return True
    return False

def draw_move(board, free_fields):
    # Computer randomly selects a move from the available free fields and updates the board
    if not free_fields:
        return board
    row, col = choice(free_fields)
    board[row][col] = 'X'
    return board

def main():
    board = [[1, 2, 3], [4, 'X', 6], [7, 8, 9]]  # Initial board setup
    display_board(board)  # Display the initial board

    while True:
        free_fields = make_list_of_free_fields(board)  # Get current free fields
        board = enter_move(board, free_fields)  # Player makes a move

        # Check for a player win
        if victory_for(board, 'O'):
            display_board(board)
            print("You won!")
            break

        # Computer's turn
        free_fields = make_list_of_free_fields(board)  # Update free fields after player's move
        board = draw_move(board, free_fields)  # Computer makes a move
        display_board(board)

        # Check for a computer win
        if victory_for(board, 'X'):
            display_board(board)
            print("Computer won!")
            break

        # Check for a tie
        if not make_list_of_free_fields(board):
            print("It's a tie!")
            break

if __name__ == "__main__":
    print("launch as a script")
    main()
else :
    print("launched as a module")
````

## Exercise 9 — Seven-Segment Display {#ex-exercise-9-seven-segment-display}

**Estimated Time:** 45 minutes  
**Level of Difficulty:** Hard  

**Scenario**  
You’ve surely seen a seven-segment display: a device that shows one decimal digit by lighting a subset of seven segments.  
In this exercise, you’ll **simulate** such a display using **single LEDs** (characters) instead of physical segments.

Each digit is built from **13 LEDs** arranged on **5 text rows** (some lit, some dark). Here’s how we imagine the output:

```
  # ### ### # # ### ### ### ### ### ### 
  #   #   # # # #   #     # # # # # # # 
  # ### ### ### ### ###   # ### ### # # 
  # #     #   #   # # #   # # #   # # # 
  # ### ###   # ### ###   # ### ### ###
```

> Note: digit **8** lights all LEDs.

**Task**  
Write a function `print_number(num)` that prints any **non-negative integer** using the LED layout above.

::: {.callout-tip collapse="true" title="Show hint"}
**Tip**  
Use a list (or dict) of 10 patterns (one per digit). For example:
:::

```python
digits = [
    '1111110',  # 0
    '0110000',  # 1
    '1101101',  # 2
    '1111001',  # 3
    '0110011',  # 4
    '1011011',  # 5
    '1011111',  # 6
    '1110000',  # 7
    '1111111',  # 8
    '1111011',  # 9
]
```

::: {.callout-tip collapse="true" title="Show hint"}
**Hints:**
:::

- Think in 5 text rows × 3 columns per digit, then concatenate across digits.
- Map each of the 7 segments (indexes 0–6 in the pattern string) to the LED positions you want to turn on.
- Build each row as a string and print at the end to avoid flicker/misalignment.

````python
# Define a dictionary that maps each digit character to its corresponding pattern
digit_patterns = {
    '0': '1111110',
    '1': '0110000',
    '2': '1101101',
    '3': '1111001',
    '4': '0110011',
    '5': '1011011',
    '6': '1011111',
    '7': '1110000',
    '8': '1111111',
    '9': '1111011',
}

def print_number(num):
    digs = str(num)
    lines = ['' for _ in range(5)]
    
    for d in digs:
        # Look up the pattern for the current digit in the dictionary
        ptrn = digit_patterns.get(d, '       ')  # Default to all spaces if the digit is not in the dictionary
        segs = [[' ', ' ', ' '] for _ in range(5)]
        
        if ptrn[0] == '1':
            segs[0][0] = segs[0][1] = segs[0][2] = '#'
        if ptrn[1] == '1':
            segs[0][2] = segs[1][2] = segs[2][2] = '#'
        if ptrn[2] == '1':
            segs[2][2] = segs[3][2] = segs[4][2] = '#'
        if ptrn[3] == '1':
            segs[4][0] = segs[4][1] = segs[4][2] = '#'
        if ptrn[4] == '1':
            segs[2][0] = segs[3][0] = segs[4][0] = '#'
        if ptrn[5] == '1':
            segs[0][0] = segs[1][0] = segs[2][0] = '#'
        if ptrn[6] == '1':
            segs[2][0] = segs[2][1] = segs[2][2] = '#'
        
        for lin in range(5):
            lines[lin] += ''.join(segs[lin]) + ' '
    
    for lin in lines:
        print(lin)

print_number(int(input("Enter the number you wish to display: ")))


````
