20. Now we would like to investigate the relationship between attrition and education field. We can't proceed like the previous analyses, as attrition and education field are both categorical variables! Let's do something else:
* Convert the Attrition column such that "Yes" is mapped to 1 and "No" is mapped to 0
* Then, compute the average of Attrition grouped by the education field
* Finally, plot the average attrition as a function of the education field.
* What is the education field with the highest attrition rate?

```python
df["Attrition"] = df["Attrition"].map({"Yes": 1, "No": 0})
df[["Attrition"]].head()
```

```python
df_attrition = df.groupby("EducationField")["Attrition"].mean().reset_index()
df_attrition
```

```python
fig = px.bar(df_attrition, x="EducationField", y="Attrition",
             title="Attrition rate by education field",
             labels={"Attrition": "Attrition rate"})
fig.show()
```

Human Resources and Technical Degree have the highest attrition rates.
