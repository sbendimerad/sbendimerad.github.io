1. In the end we decided to build this exercise around CoinGecko, can you give us at least one reason why we made that decision? If you would like to go further you can list some advantages and drawbacks for each of these APIs.

* **Coinbase**: 
    * 10 000 requests per hour
    * does not require authentication for price data
    * Only able to query one date and one crypto at a time
* **Coinlib**:
    * Between 60 and 180 requests per hour
    * No historical data
    * Need API key
* **CoinPaprika**:
    * 36 000 requests per hour
    * historical data limited on free plan
    * No authentication required
* **BraveNewCoin**:
    * API key + authentication token valid for 24hours
    * 200 000 requests per month
* **CoinGecko**:
    * No API key needed
    * Rate limit of 50 calls/minute
    * Possibility to get historical data and OHLC (Open High Low Close) prices

3. Call the API to get the list of the first *250* crypto-currencies in terms of market cap :
* Response result should be ordered by market cap in descending order.
* **vs_currency** parameter should be set to *usd*
* Check that the status code is equal to 200

````python
import requests

url_base = 'https://api.coingecko.com/api/v3/'
url_coins = 'coins/'
url_markets = 'markets/'
````

````python
payload = {'vs_currency':'usd', 'order':'market_cap_desc', 'per_page':'250'}
coins = requests.get(f"{url_base}/{url_coins}/{url_markets}", params = payload)


print('status code:', coins, '\n\n')
````

10. You have already:
- **Question 5**: Retrieved the list of the top 10 cryptocurrencies (with their IDs).  
- **Question 7**: Extracted the OHLC data for Bitcoin and converted timestamps to readable dates.  

Now, extend this work to the top 10 cryptos.

* From the result of **Question 5**, create a Python list called `coins_id` containing the IDs of the top 10 cryptocurrencies.  
   *(Example: `["bitcoin", "ethereum", "tether", ...]`)*

* For each cryptocurrency in `coins_id`, request the **OHLC data for the last 365 days** using the CoinGecko API (similar to what you did for Bitcoin in Question 7).  
   - Endpoint: `https://api.coingecko.com/api/v3/coins/{coin_id}/ohlc`  
   - Use the parameters:  
     ```python
     payload = {'vs_currency': 'usd', 'days': '365'}
     ```

* Store the results in a dictionary called `ohlc_data`, where the **keys** are the crypto IDs and the **values** are their OHLC data.

* Convert each crypto’s OHLC data into a **pandas DataFrame** with columns:  
   `["date", "open", "high", "low", "close"]`. Dont forget to convert the timestamp.


````python
coins_id = []
for coin in first_ten_coins:
    coins_id.append(coin['id']) 
````

````python
coins_id
````

````python
payload = {'vs_currency':'usd', 'days':'365'}

ohlc_data = {}

for coin_id in coins_id:  # loop over the list of ids
    response = requests.get(
        f'https://api.coingecko.com/api/v3/coins/{coin_id}/ohlc',
        params=payload
    )
    ohlc_data[coin_id] = response.json()  # save data
    print(f"✅ Got OHLC for {coin_id}")
````

````python
# Example: access Bitcoin data
ohlc_data['bitcoin'][:3]  # show first 3 rows
````

````python
all_dfs = []

for coin_id, data in ohlc_data.items():
    # Create DataFrame
    df = pd.DataFrame(data, columns=["timestamp_ms", "open", "high", "low", "close"])
    
    # Convert ms → datetime (YYYY-MM-DD)
    df["date"] = pd.to_datetime(df["timestamp_ms"], unit="ms").dt.date
    
    # Keep only date + OHLC
    df = df[["date", "open", "high", "low", "close"]]
    
    # Add coin name
    df["coin"] = coin_id
    
    all_dfs.append(df)

# Merge into one DataFrame
top10_df = pd.concat(all_dfs, ignore_index=True)
````
