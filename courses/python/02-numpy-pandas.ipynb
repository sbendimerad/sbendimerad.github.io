#### Initialize numpy arrays with pre-defined values

##### Create matrices of 0s or 1s

It is useful to be able to create matrices composed only of 0 or 1. To do this, you can use: ```np.zeros()``` or ```np.ones()```.

````python
# Initialization to 0 (matrix 3x4)
display(np.zeros((3,4)))

# Initialization to 1 (matrix 5x4)
display(np.ones((5,4)))

# Third order tensor (can be seen as 2 matrices 3x4)
np.ones((2,3,4))
````

##### Create a list of regularly spaced values
We have seen that the `range()` function in python can generate lists of regularly spaced integers. In the same way, the np.arange() method allows to generate lists of values (including decimal numbers):

````python
# Generates a list of numbers from 3.2 to 4.8 excluded with a step of 0.2.
np.arange(3.2, 4.8, 0.2)
````

#### Initialize an array from a user-defined function

````python
def f(x, y):
    return 10*x+y

# Creating an array from the function from above
np.fromfunction(f,(5,4),dtype=int)
````

### Accessing items in a multidimensional array

````python
# Creation of a matrix
c = np.array([[  0,  1,  2],
               [ 10, 12, 13],
               [100,101,102],
               [110,112,113]])
````

````python
c[3, 2] # Accessing to the value contained in the 4th row, 3rd column, is accessed
````

````python
# Creation of a third order tensor
c = np.array( [[[  0,  1,  2],
                 [ 10, 12, 13]],
                [[100,101,102],
                 [110,112,113]]])
````

````python
c[0, 1, 2] # We access the value contained in the 2nd row, 3rd column of the first matrix
````

#### Iterate over numpy arrays

````python
# Create matrix
c = np.array([[  0,  1,  2],
               [ 10, 12, 13],
               [100,101,102],
               [110,112,113]])
````

##### Iterate on arrays

You can use the `np.flat` attribute to easily iterate across all the elements in an array:

````python
# Iterate on the elements of the matrix

# Not recommanded (may be slow):
for row in c:
    for e in row:
        print(e)

print()
print('----------')
print()

# Better:
for e in c.flat:
    print(e)
````

#### Slices
In the same way as in pandas, you can use the slices syntax to select a sub-part of the arrays.

````python
def f(x, y):
    return 10*x+y

# Creating an array from the function from above
b = np.fromfunction(f,(5,4),dtype=int)
print(b)
````

````python
# Slices
print(b[0:5, 1])  # every lines of the second column
print()
print(b[:, 1])  # every lines of the second column

print()
print('----------')
print()

print(b[1:3, : ]) # 2nd and 3rd lines of all columns

print()
print('----------')
print()

print(b[2, 1:3]) # 3rd line, 2nd and 3rd columns

print()
print('----------')
print()

print(b[2, 2:4]) # 3rd line, 2 last columns
print()
print(b[2, -2:]) # 3rd line, 2 last columns
````

#### Masks
Similarly, you can also use masks to select elements from your arrays based on a condition:

````python
# Create a table
a = np.arange(12).reshape(3,4)
display(a)
# Create a mask
b = a > 4
b
````

````python
# Use the mask to select elements of the matrix
a[b]
````

Only the values for which the mask had the value `True` were shown.

### Manipulation of the shape of a numpy array

Finally, it is good to talk about the *shape* of an array. This is especially useful when manipulating images when you're attacking deep learning.

#### What is the shape of a numpy array?

The *shape* of an array gives us information about its dimension and the number of elements present in each dimension.

````python
# This is the shape of a matrix

c = np.array([[  0,  1,  2],
               [ 10, 12, 13],
               [100,101,102],
               [110,112,113]])

c.shape
````

````python
# A matrix is an array of the second dimension (order):
len(c.shape)
````

````python
# A vector is an array of the first dimension:
v = np.array([1,2,3,4])
len(v.shape)
````

#### Shape & Reshape

Quite often you need to change the shape of a matrix. For example, you might want to invert the rows and columns of your matrix or *flatten* your matrix. You can do this with ``.reshape()```

````python
# Change array's shape with reshape()
c = np.array([[  0,  1,  2],
               [ 10, 12, 13],
               [100,101,102],
               [110,112,113]])

c.reshape((2,6)) # Be careful: n_lines x n_column must remain equal to the number of elements present in c
````

You can also *guess* the value inside one of the dimensions of your array by setting the value `-1`.

````python
# You can use -1 to let numpy guess the number of rows or the number of columns:
c.reshape((-1,4))
````

````python
c.reshape((2,-1))
````

#### Combine databases

````python
# Next, we will manipulate the following databases:
data_sales1 = {
    'sales_id' : ['001','002','003','004'],
    'people_id' : [1, 4, 2, 1],
    'product_id' : ['X789', 'X999', 'X789', 'X990']
              }

data_sales2 = {
    'sales_id' : ['005', '006', '007'],
    'people_id' : [0, 3, 2],
    'product_id' : ['X789', 'X999', 'X789']
               }

data_products = {
    'product_id' : ['X789', 'X999', 'X990'],
    'product_desc' : ['Apple', 'Banana', 'Orange']
                }

data_people = {
    'id' : [0, 1, 2, 3, 5],
    'name' : ['Paul','Perrine','Moussa','Michel', 'Anne'],
    'age' : [None, 67, 24, 76, 47]
              }

df_sales1 = pd.DataFrame(data_sales1)
df_sales2 = pd.DataFrame(data_sales2)
df_products = pd.DataFrame(data_products)
df_people = pd.DataFrame(data_people)

display(df_sales1)
display(df_sales2)
display(df_products)
display(df_people)
````

##### Concatenate lines

![concat: the rows of the second table are added under the first](img/02-numpy-pandas/concat_rows.svg)

````python
# Concatenate the lines of df_sales2 after df_sales1 :
df_sales = pd.concat([df_sales1,df_sales2], ignore_index=True)
display(df_sales)
````

##### Join operations

![merge: rows are matched on a key column](img/02-numpy-pandas/merge_on_key.svg)

````python
# Join columns from sales and product
# Simple case: All product_id in df_sales are present in df_product and vice versa

df_sales_product = df_sales.merge(df_products,on='product_id')
display(df_sales_product)
````

````python
df_final1 = df_sales_product.merge(df_people, left_on='people_id', right_on='id', how='inner')
````

````python
# Join columns from people with sales_product
# Warning 1: the columns containing the join id do not have the same name
# Warning 2: the people_id 4 present in df_sales_product does not exist in df_people!
# People_id 5 is in df_people but not in df_sales_product
# We have to make a choice: what to do with the missing ids?

# Inner join: we only keep the ids that are in both tables
df_final1 = df_sales_product.merge(df_people, left_on='people_id', right_on='id', how='left')

# Outer join: we keep all the ids we find in both tables and fill the missing values in with NaN
df_final2 = df_sales_product.merge(df_people, left_on='people_id', right_on='id', how='outer')

# Left join: we keep all the ids from the left table and fill the mising values with NaN
df_final3 = df_sales_product.merge(df_people, left_on='people_id', right_on='id', how='left')
````

## Dealing with datetimes and strings 🏂

`Datetime` and `str` are very widespread python types:

* [datetime](https://docs.python.org/3/library/datetime.html)
* [str](https://docs.python.org/3/library/string.html)

Manipulating that kind of data may reveal difficult, but fortunately there exists many tools that will help you master it. In particular, pandas provides the `.dt` and `.str` submodules that will help you make basic operations on datetime and string objects.

In this lecture, you will learn how to:

* Convert strings to datetimes and the way back
* Extract features from datetimes
* Make operations on datetimes thanks to the Timedelta type
* Make basic operations on strings
* Extract features from strings

````python
import pandas as pd
````

### Pandas and datetime : the .dt submodule 📅

Pandas provides very advanced datetime-handling features. In this section, some basic attributes/methods from pandas' `.dt` submodule are introduced.

Let's begin by reading some data containing dates and times:

````python
df = pd.read_csv("../data/datetime_df.csv")
df.head()
````

````python
df.dtypes
````

#### Date+time = Datetime

##### Conversion : from string to datetime
Very often, when you load data into a pandas DataFrame or Series, datetime formats will be read as strings. The first thing to do to avoid compatibility issues is to use `pd.to_datetime` to convert the Series/column to the right type:

````python
df['dt'] = pd.to_datetime(df["dt"])
df.dtypes
````

##### Feature extraction
The `.dt` submodule contains many attributes that allow to extract features very easily from the `datetime` objects. Below are the most basic features, but there exist a lot of other attributes, don't hesitate to look for what you need in [pandas API](https://pandas.pydata.org/docs/reference/api/pandas.Series.dt.date.html) 🤓

````python
df['year'] = df['dt'].dt.year
df['month'] = df['dt'].dt.month
df['day'] = df['dt'].dt.day
df['weekday'] = df['dt'].dt.dayofweek
df['hour'] = df['dt'].dt.hour
df['minute'] = df['dt'].dt.minute
df['second'] = df['dt'].dt.second
````

````python
df.head()
````

#### Timedelta

When dealing with dates and times, another useful type is python's [timedelta](https://docs.python.org/3/library/datetime.html#timedelta-objects). `Timedelta` objects are useful to make some computations on `datetime` objects, for example: computing the time elapsed between two events, adding 24 hours to a datetime, etc...

Pandas `.dt` submodule also includes `timedelta` management:

````python
# Create an object representing a timedelta of 1 day
days_1 = pd.to_timedelta(1, unit = 'd')
# Use the timedelta object to apply a systematic shift of - 1 day to all datetimes:
df["dt_1d"] = df["dt"] - days_1

# Create an object representing a timedelta of 10 hours
hours_10 = pd.to_timedelta(10, unit = 'h')
# Use the timedelta object to apply a systematic shift of + 10 hours to all datetimes:
df["dt_10h"] = df["dt"] + hours_10

df.head()
````

````python
# It's possible to use the operator "-"" between two datetime Series
# it will compute the time elapsed between the two Series and return it as a Timedelta
elapsed_time = df["dt_10h"] - df["dt"]
print("Elapsed time between dt and dt_10h:")
elapsed_time
````

````python
# It's possible to convert this information into a duration in days or seconds:
print("Elapsed time in days:")
print(elapsed_time.dt.days)
print()
print("Elapsed time in seconds:")
print(elapsed_time.dt.seconds)
````

### Pandas and strings: the .str submodule 📔
Similarly, pandas provides a `.str` submodule that allows to manipulate Series containing strings.

Let's begin by reading a file containing string data:


````python
df = pd.read_csv("../data/string_df.csv")
df.head()
````

#### Basic transformations

````python
# Lowercase
df["description"].str.lower()
````

````python
# Replace each occurrence of a pattern
df["description"].str.replace(" and ", "&") # pattern matching is case-sensitive and takes whitespaces into account
````

````python
# Slicing
print(df["description"].str.slice(start=3)) # remove 3 first characters
print()
print(df["description"].str.slice(start=0, stop=10)) # Keep only 10 first characters
````

#### Feature extraction

````python
# Extract string length
df["desc_len"] = df["description"].str.len()

# Test if string starts with a given pattern
df["desc_starts_with_This"] = df["description"].str.startswith("This")

# Test if string contains in a given pattern
df["desc_contains_This"] = df["description"].str.contains("This")

# Count the occurrences of a pattern
df["desc_count_This"] = df["description"].str.count("This")

# Split string into a list of strings, by using a given separator
df["desc_split"] = df["description"].str.split(",")

# For the braves: how to count words
df["desc_wordcount"] = df['description'].str.split(" ").apply(lambda x: len(x))

df.head()
````

With these few features from pandas `.dt` and `.str` submodules, you'll be able to carry on some basic analysis on datetimes and string data 🥳

You will learn more advanced ways of extracting features from unstructured data later 😉
