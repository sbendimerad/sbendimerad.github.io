## Company description 📇

Uber is a ride-hailing company: through its app, users book a ride with a driver nearby. It started as an alternative to taxis and now also delivers food and packages, in about 70 countries.

## Project 🚧

A recurring problem for Uber's team: drivers are not always where users need them. A user may be waiting in a busy business district while drivers look for customers in a quiet neighbourhood a few kilometres away. The user then waits 10 to 15 minutes, while Uber's research shows that most users accept to wait 5 to 7 minutes before cancelling.

So the data team would like the driver app to **recommend hot zones**: the places where drivers should be at a given time of the day and day of the week.

## Goals 🎯

Uber has the location of past pickups. Your objective is to:

- build an algorithm that finds hot zones from past pickups;
- show the results on maps that a driver (or a manager) can read at a glance.

## Scope of this project 🖼️

Uber wants to test the feature in **New York City** first. The data contains every Uber pickup in New York in **April 2014** (564,516 rides), with the date and time and the GPS position of each pickup. It was obtained by FiveThirtyEight from the NYC Taxi & Limousine Commission.

| Column | Meaning |
|---|---|
| `Date/Time` | date and time of the pickup |
| `Lat`, `Lon` | latitude and longitude of the pickup |
| `Base` | code of the Uber base the car was affiliated to (not needed here) |

## Helpers 🦮

### Start small, then generalize

Uber wants hot zones per hour and per day of the week, but don't start with everything. **Pick one day and one hour** (for example Monday, 5 pm), find a method that gives sensible zones, and only then turn it into a function that you can run for any day and hour.

### Clustering is your friend

The pickups can be grouped into clusters: the **centre of a cluster** pins a hot zone, and its **number of pickups** says how hot it is.

- **K-Means** splits *every* pickup into $k$ groups. Think about what its zones mean: is every zone "hot"?
- **DBSCAN** only keeps the places where pickups are **dense** and labels the rest as noise. Latitude and longitude are angles: use `metric="haversine"` on coordinates in **radians**, so that `eps` is a distance on the Earth (`eps = metres / 6371000`). The density of pickups is very different in Manhattan and in the outer boroughs: try several values of `eps` and `min_samples`, and look at the result on a map each time.
- The number of pickups changes a lot between days and hours. When you generalize, should `min_samples` stay fixed, or follow the number of pickups in the time slot?

### Maps with plotly

`px.scatter_map(df, lat="Lat", lon="Lon", color=..., size=..., zoom=10, map_style="carto-positron")` draws points on a map. With `animation_frame="hour"`, you get a slider that shows the zones hour by hour: a nice dashboard!

With more than half a million rows, plot **samples** (or only the cluster centres), not all the points.
