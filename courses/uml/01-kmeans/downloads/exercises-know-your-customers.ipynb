## The dataset 🧾

`data/mall_customers.csv` is a small teaching dataset of 200 fictional customers. For each customer, it gives:

* `Genre`: the gender (`Male` / `Female`);
* `Age`;
* `Annual Income (k$)`: the annual income, in thousands of dollars;
