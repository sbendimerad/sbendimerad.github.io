## Company description 📇

Cairn Outdoor is an outdoor brand (a fictional one, created for this course). It sells clothing, footwear, backpacks and camping gear for hiking, climbing, running and winter sports, mostly through its online shop.

## Project 🚧

The marketing team wants to use machine learning to boost online sales. They have two ideas:

- A **recommender system**: on each product page, a section "You might also like..." suggests products that are similar to the one the customer is looking at.
- A better **catalogue structure**: the website menu has 7 categories (Jackets, Tops, Bottoms, Bags, Camping, Footwear, Accessories). Are they the right ones? Can unsupervised methods suggest **new categories**, or new ways to browse the shop?

## Goals 🎯

The project has three steps:

1. Find groups of products whose descriptions are similar (clustering).
2. Use these groups to build a simple recommender system.
3. Use topic modeling to find the latent topics of the descriptions.

## Scope of this project 🖼️

The file `data/cairn_outdoor_catalog.csv` contains the 304 products of the catalogue:

| Column | Meaning |
|---|---|
| `id` | product id |
| `name` | product name |
| `description` | the text of the product page |
| `category` | the category of the website menu |

## Deliverable 📬

To complete this project, you should:

- train at least one **clustering model** on the descriptions, and describe its clusters with word clouds;
- write a function that takes the id of a product and returns a list of **similar products**;
- train at least one **TruncatedSVD** model on the descriptions and describe its topics with word clouds;
- compare your clusters and topics with the current categories, and make a suggestion to the marketing team.

## Helpers 🦮

### Preprocessing the texts

- Clean the descriptions: lowercase, remove punctuation and numbers, remove stop words and lemmatize (`spacy`, model `en_core_web_sm`). See the NLP section.
- Encode the texts with **TF-IDF** (`TfidfVectorizer`).

### Part 1: groups of similar products

- Train `DBSCAN` on the TF-IDF matrix. For texts, use `metric="cosine"` instead of the Euclidean distance (see the NLP section: cosine compares the *direction* of two documents, not their length).
- Try several values of `eps` and `min_samples`: aim for about 10 to 20 clusters and not too many outliers.
- Display a **word cloud** for each cluster: do the clusters contain different words? Compare the clusters with the `category` column with `pd.crosstab`.

### Part 2: recommender system

- Write a function `find_similar_items(item_id, n=5)` that returns `n` products of the **same cluster** as `item_id`.
- Improve it: rank the products of the cluster by **cosine similarity** with the chosen product. What should the function do for a product labelled as noise (cluster -1)?

### Part 3: topic modeling

- Train a `TruncatedSVD` on the TF-IDF matrix (LSA, see the topic modeling section), for example with 5 to 8 topics.
- For each topic, show the words with the largest weights in a word cloud. Are the topics the same thing as the clusters?
