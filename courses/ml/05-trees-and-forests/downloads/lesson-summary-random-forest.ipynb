### The essentials


## Bagging

Bagging means training several models on the same prediction problem and combining their strengths to improve the results. In general, the diversity of the models gives better predictions while limiting the variance, so the final model generalizes well to the validation data.


## Bootstrap

Bootstrapping means creating new datasets from an existing dataset by drawing observations from it with replacement. This randomly generates various datasets that resemble the original data without being identical to it. Models can then be trained on varied data, each dataset randomly representing different characteristics of the original one.

The Bootstrap on the columns consists in using only a random subset of the explanatory variables when building the models of the Bagging (in a Random Forest, a new subset is drawn at each split). This encourages the final model to capture different types of patterns, by focusing on different variables.


## Random Forest

Random Forests are versatile models, suitable for both classification and regression problems. A Random Forest is a set of decision trees combined by bagging.

The main advantages of Random Forests: 
* no assumption is needed about the shape of the function f that links the explanatory variables to the target variable;
* the model is quick to train;
* many hyperparameters let you control the bias/variance balance.

The drawbacks come from the number of hyperparameters that need to be tuned to find the best model. The final model is also difficult to interpret.
