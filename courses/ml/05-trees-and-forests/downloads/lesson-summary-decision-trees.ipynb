## Decision trees 🌳🌳


### The essentials


## Successive train splits

Decision trees are models that split the dataset again and again to produce their predictions. At each division step, the model selects the variable and the threshold value that give the two sub-datasets in which the target variable is as homogeneous as possible.


## Control of bias and variance

Decision trees naturally stop dividing the dataset when all branches contain sub-datasets in which the target variable is homogeneous (the same value for all observations). This gives accurate predictions on the _train_ data, but also gives the model a very high variance. To reduce the variance, we can impose constraints on the model, such as a minimum number of observations in the leaves, or a maximum depth of k successive divisions.


## Advantages / disadvantages

The main advantage of the decision tree is that it combines simplicity (it is basically a set of binary rules: X1 smaller or larger than a value...) and good performance. Indeed, it can fit the training data as closely as we want. Its major drawback is that it is not a continuous model: the predictions take the form of a step function. This is not a big problem for classification, but it is a bit annoying for regression. You can also picture the model as a grid over the space of the explanatory variables, which does not suit every kind of problem.
