In this project, you take part in a machine learning competition like the ones organised on [Kaggle](https://www.kaggle.com/). You work in Jupyter notebooks as usual, but in the end you **submit your model's predictions to your teacher**, who scores them in an independent way. The scores of all the teams go into a **leaderboard** 🏅🏅

## How a machine learning challenge works 🚴🚴

- The data comes in two files:
    - `conversion_train.csv` contains **labelled data**: both the explanatory variables X and the target Y. You use it as usual: train/validation split, preprocessing, performance assessment, different models, hyperparameter tuning…
    - `conversion_test.csv` contains **new examples** in the same format, but **without the target**. Once your model is trained, you predict these rows and send your predictions. The organisers compare them with the true labels, which only they have. Nobody can cheat 🤸
- Everybody knows in advance **which metric** decides the ranking: make sure you use the same one to evaluate your own models!

## The company 📇

*The Weekly Model* is a newsletter written by independent data scientists (the newsletter and its data are fictional, made for this course). Anyone can leave their email address on the website to receive the week's news about data science and its applications.

## The project 🚧

The team would like to understand the behaviour of the people who visit their website. Can a model predict whether a visitor will **subscribe** to the newsletter (a *conversion*), using only a few pieces of information about them? They would also like to read the model, to find which features explain subscriptions, and maybe discover a **lever for action** to improve the conversion rate.

So they organised a competition and shared a dataset about the traffic on their website. Teams are ranked with the **F1-score**.

| Column | Meaning |
|---|---|
| `country` | country of the visitor |
| `age` | age of the visitor |
| `new_user` | 1 if this is the visitor's first visit, else 0 |
| `source` | how the visitor arrived: `Seo` (search engine), `Ads` (advertising) or `Direct` |
| `total_pages_visited` | number of pages read during the visit |
| `converted` | **target**: 1 if the visitor subscribed, else 0 |

## Goals 🎯

The project has four parts:

1. **Explore and build a baseline**: an EDA, the preprocessing, and a first model trained on `conversion_train.csv`.
2. **Improve your F1-score** on your own validation set: feature engineering, feature selection, regularisation, non-linear models, hyperparameter tuning with a grid search, the decision threshold…
3. **Submit**: when you are happy with your model, predict `conversion_test.csv` and save the predictions in a CSV file for your teacher. You can submit as many times as you like: try different models!
4. **Interpret**: analyse your best model. Is there a lever for action to improve the conversion rate? What would you recommend to the team?

## Deliverables 📬

Your team should:

- create relevant figures for the EDA;
- train at least one model that predicts conversions, and evaluate it (F1-score, confusion matrix);
- make at least one submission to the leaderboard;
- analyse your best model and make recommendations to improve the conversion rate.

### Submission format

A CSV file named `conversion_predictions_<team-name>.csv` (for example `conversion_predictions_red-pandas.csv`), with:

- **one column** named `converted`, containing 0 or 1;
- **one row per row** of `conversion_test.csv`, in the same order;
- **no index** (`to_csv(..., index=False)`).

## Tips 🦮

- Conversions are **rare**: only a few percent of visitors subscribe. A model that always predicts 0 is right more than 95% of the time and still useless. That is why the ranking uses the F1-score, which looks at the subscribers you find (recall) and at the false alarms (precision).
- The dataset has close to 300,000 rows: work on a **sample** for your figures.
- The **template** (`conversion_rate_template.ipynb`) implements every step for a very simple model (a logistic regression on a single feature). Start from it and improve each step. May the force be with you 🧨🧨
