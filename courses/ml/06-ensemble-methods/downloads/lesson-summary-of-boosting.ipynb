## Boosting


## The essentials


## General principle

In the literature, boosting is often said to turn a _weak learner_ into a _strong learner_, i.e. to turn a model that is only slightly better than chance into a model that achieves the best possible performance on the data in question.


## Algorithm

Boosting starts by training the selected model on the training data and stores it in memory. Then it computes predictions for each observation. AdaBoost then assigns a weight to each observation according to the prediction errors: it gives a higher weight to the observations it got wrong, and a lower weight to the ones it predicted well. The same model is trained again on the weighted data and kept in memory. This process is repeated several times. The result is a sequence of models that focus on different groups of observations, based on the previous performance. These models are aggregated into the final model, which in theory should perform better on the data than the original model.
Another boosting method, gradient boosting, trains the sequence of models differently: the first model of the sequence is trained to predict $y$, the target variable, and every following model is trained on the residuals of the previous model in the sequence. The final model is the sum of all the elementary models trained in the sequence.


## Methodology

The essential prerequisite for boosting to work is to find a suitable base model for the Machine Learning problem at hand. The base model used in boosting should have a relatively low variance and a high bias. This is just a rule of thumb, but boosting algorithms in Python usually use decision trees by default. Boosting can be tuned by changing the number of models to train and the learning rate (which controls how fast the weights of the observations change: if it is too low, boosting will take too long; if it is too high, the models may behave chaotically). We are often limited in the number of models by computing time; on the other hand, we can find an optimal learning rate with a grid search. A reasonable learning rate should be less than 1.
