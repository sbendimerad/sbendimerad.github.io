#### Preprocessing with pandas

**Drop the rows where the target is missing:**
 - Here, the target variable (Y) corresponds to the column *Weekly_Sales*. This column has some missing values.
 - We never impute the target: it could bias the predictions!
 - So we simply drop the rows where *Weekly_Sales* is missing.

**Create usable features from the *Date* column:**
The *Date* column cannot be used as it is in the model. You can either drop it, or create new columns with the following numeric features:
- *year*
- *month*
- *day*

(All the dates are Fridays, the end of each sales week: a day-of-week feature would always have the same value, so it would be useless.)

⚠️ The dates are written day first (`18-02-2011`): tell pandas with `pd.to_datetime(..., format="%d-%m-%Y")`, otherwise some dates are read month first and silently become wrong.

**Drop the rows containing invalid values or outliers:**
In this project, a value is considered an outlier if it falls outside the range $[\bar{X} - 3\sigma, \bar{X} + 3\sigma]$ of its column. This concerns the columns *Temperature*, *Fuel_Price*, *CPI* and *Unemployment*.



**Target variable (Y) that we will try to predict, to separate from the others:** *Weekly_Sales*
